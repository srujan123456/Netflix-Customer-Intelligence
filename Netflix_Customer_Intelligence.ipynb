```python
import pandas as pd
import numpy as np

# Create a separate cleaned copy; do not modify the source DataFrame
payments_clean = payments.copy()

# Ensure payment and subscription dates are in datetime format
payments_clean['payment_date'] = pd.to_datetime(payments_clean['payment_date'], errors='coerce')

subscriptions_for_matching = subscriptions_clean.copy()

date_columns = ['subscription_start_date','subscription_end_date']

for column in date_columns:
    subscriptions_for_matching[column] = pd.to_datetime(subscriptions_for_matching[column], errors='coerce')

# Keep only the columns required for reconciliation
payment_columns = [
    'payment_id',
    'customer_id',
    'payment_date'
]

subscription_columns = [
    'subscription_id',
    'customer_id',
    'subscription_start_date',
    'subscription_end_date',
    'subscription_status'
]

# Match payments to subscription records for the same customer
candidates = payments_clean[payment_columns].merge(subscriptions_for_matching[subscription_columns],on='customer_id',
    how='left')

# A payment must occur on or after the subscription start date
after_start = (candidates['payment_date']>= candidates['subscription_start_date'])

# A payment must be on or before the subscription end date.
# An Active subscription with no end date is treated as ongoing.
within_end = ((candidates['subscription_end_date'].notna()& (candidates['payment_date']<= candidates['subscription_end_date']
        )
    )
    |
    (candidates['subscription_status'].eq('Active')& candidates['subscription_end_date'].isna()))

# Retain only candidates with valid date-window matches
matches = candidates.loc[after_start & within_end,['payment_id', 'subscription_id']].drop_duplicates()

# Count the number of matching subscriptions for each payment
match_count = matches.groupby('payment_id')['subscription_id'].nunique()

# Assign a subscription only when exactly one match exists
unique_matches = matches[matches['payment_id'].map(match_count).eq(1)]

unique_subscription_map = (unique_matches.drop_duplicates('payment_id').set_index('payment_id')['subscription_id'])

# Fill subscription_id in the cleaned payment data
# Payments with zero or multiple matches remain blank
payments_clean['subscription_id'] = (payments_clean['payment_id'].map(unique_subscription_map).astype('string'))

# Create an audit table so we can review unresolved payments
subscription_match_audit = payments_clean[['payment_id', 'customer_id', 'payment_date']].copy()

subscription_match_audit['match_count'] = (subscription_match_audit['payment_id'].map(match_count).fillna(0).astype(int))

subscription_match_audit['match_status'] = np.select(
    [
        subscription_match_audit['match_count'].eq(1),
        subscription_match_audit['match_count'].gt(1)
    ],
    [
        'Matched',
        'Ambiguous'
    ],
    default='No matching subscription window'
)

# Display reconciliation results
print("Reconciliation status:")
print(subscription_match_audit['match_status'].value_counts())

print("\nPayments with a subscription assigned:",payments_clean['subscription_id'].notna().sum())

print("Payments still missing subscription_id:",payments_clean['subscription_id'].isna().sum())
```

```python
import pandas as pd
import numpy as np

# Select only payments that have no matching subscription window
unmatched_no_window = subscription_match_audit[
    subscription_match_audit['match_status']
    == 'No matching subscription window'
].copy()

# Create a separate copy of the subscription data for analysis
subs_for_analysis = subscriptions_clean.copy()

# Ensure subscription dates use datetime format
date_columns = [
    'subscription_start_date',
    'subscription_end_date'
]

for column in date_columns:
    subs_for_analysis[column] = pd.to_datetime(
        subs_for_analysis[column],
        errors='coerce'
    )

# Group subscriptions by customer for easier investigation
subscriptions_by_customer = {
    customer_id: group.sort_values('subscription_start_date')
    for customer_id, group in subs_for_analysis.groupby('customer_id')
}


# Function to classify each unmatched payment
def classify_unmatched_payment(payment):

    customer_id = payment['customer_id']
    payment_date = pd.to_datetime(
        payment['payment_date'],
        errors='coerce'
    )

    # Check whether the payment date is available
    if pd.isna(payment_date):
        return 'Missing payment date'

    # Get all subscriptions belonging to this customer
    customer_subs = subscriptions_by_customer.get(
        customer_id,
        pd.DataFrame()
    )

    # Category 1: Customer has no subscription record
    if customer_subs.empty:
        return 'No subscription record for customer'

    starts = customer_subs['subscription_start_date']
    ends = customer_subs['subscription_end_date']
    statuses = customer_subs['subscription_status']

    # Identify incomplete or inconsistent subscription date records
    invalid_dates = (
        starts.isna()
        |
        (
            ends.isna()
            & statuses.ne('Active')
        )
        |
        (
            starts.notna()
            & ends.notna()
            & (ends < starts)
        )
    )

    # Avoid making assumptions if the customer's subscription dates
    # contain incomplete or inconsistent records
    if invalid_dates.any():
        return 'Subscription dates incomplete/inconsistent - review'

    # Category 2: Payment occurred before the first subscription
    if payment_date < starts.min():
        return 'Payment before first recorded subscription'

    # Check whether the payment falls within any recorded window
    is_within_window = (
        (starts <= payment_date)
        &
        (
            (ends >= payment_date)
            |
            (ends.isna() & statuses.eq('Active'))
        )
    )

    if is_within_window.any():
        return 'Potential matching window - review matching logic'

    # Category 3: Payment occurred after all recorded subscriptions
    if (
        ends.notna().all()
        and payment_date > ends.max()
    ):
        return 'Payment after last recorded subscription'

    # Category 4: Payment occurred between subscription periods
    ordered_subs = customer_subs.sort_values(
        'subscription_start_date'
    )

    for i in range(1, len(ordered_subs)):

        previous_end = ordered_subs.iloc[i - 1][
            'subscription_end_date'
        ]

        next_start = ordered_subs.iloc[i][
            'subscription_start_date'
        ]

        if (
            pd.notna(previous_end)
            and previous_end < payment_date < next_start
        ):
            return 'Payment between subscription periods'

    # Category 5: No clear explanation from the available dates
    return 'Other date-window inconsistency - review'


# Apply the classification to all unmatched payments
unmatched_no_window['reason_category'] = (
    unmatched_no_window.apply(
        classify_unmatched_payment,
        axis=1
    )
)

# Display the number of payments in each category
print("Unmatched payment classification:")
display(
    unmatched_no_window['reason_category']
    .value_counts()
    .rename_axis('reason_category')
    .reset_index(name='payment_count')
)

# Verify that every unmatched payment was classified
print(
    "\nTotal unmatched payments classified:",
    len(unmatched_no_window)
)
```

```python
import pandas as pd
import numpy as np

# Work on copies so our existing cleaned datasets remain unchanged
unmatched_review = unmatched_payments[['payment_id', 'customer_id', 'payment_date']].copy()

subscriptions_review = subscriptions_clean.copy()

# Ensure all relevant dates are datetime values
date_columns = [
    'subscription_start_date',
    'subscription_end_date',
    'cancellation_date'
]

for column in date_columns:
    subscriptions_review[column] = pd.to_datetime(
        subscriptions_review[column],
        errors='coerce'
    )

unmatched_review['payment_date'] = pd.to_datetime(unmatched_review['payment_date'],errors='coerce')

# Group subscription records by customer for investigation
subscription_groups = {
    customer_id: group.sort_values('subscription_start_date')
    for customer_id, group in subscriptions_review.groupby('customer_id')
}


def classify_unmatched_payment(payment):
    # Get the payment date and customer's subscription history
    payment_date = payment['payment_date']
    customer_id = payment['customer_id']

    if pd.isna(payment_date):
        return 'Missing payment date - review'

    customer_subs = subscription_groups.get(
        customer_id,
        pd.DataFrame()
    )

    if customer_subs.empty:
        return 'No subscription record for customer'

    starts = customer_subs['subscription_start_date']
    ends = customer_subs['subscription_end_date']
    statuses = customer_subs['subscription_status']
    cancellations = customer_subs['cancellation_date']

    # Detect subscription records with incomplete or contradictory dates
    invalid_dates = (
        starts.isna()
        | (
            starts.notna()
            & ends.notna()
            & (ends < starts)
        )
        | (
            starts.notna()
            & cancellations.notna()
            & (cancellations < starts)
        )
        | (
            ends.isna()
            & statuses.ne('Active')
        )
    )

    # First check whether any documented date problem could affect
    # interpretation of the payment date
    relevant_invalid = invalid_dates & (
        starts.isna() | starts.le(payment_date)
    )

    if relevant_invalid.any():
        return 'Subscription date inconsistency - review'

    # Check whether a valid subscription window exists
    in_window = (
        starts.notna()
        & starts.le(payment_date)
        & (
            (
                ends.notna()
                & ends.ge(payment_date)
            )
            | (
                ends.isna()
                & statuses.eq('Active')
            )
        )
    )

    if in_window.any():
        return 'Potential matching window - review logic'

    # Find the earliest recorded subscription start
    valid_starts = starts.dropna()

    if not valid_starts.empty and payment_date < valid_starts.min():
        return 'Payment before first recorded subscription'

    # Check whether the payment falls between two recorded segments
    ordered = customer_subs.sort_values('subscription_start_date')

    for i in range(1, len(ordered)):
        previous_end = ordered.iloc[i - 1][
            'subscription_end_date'
        ]
        next_start = ordered.iloc[i][
            'subscription_start_date'
        ]

        if (
            pd.notna(previous_end)
            and pd.notna(next_start)
            and previous_end < payment_date < next_start
        ):
            return 'Payment between subscription periods'

    # Check for payments after the last recorded subscription window
    valid_ends = ends.dropna()

    if (
        not valid_starts.empty
        and not valid_ends.empty
        and payment_date > valid_ends.max()
        and not (
            starts.isna()
            | (
                ends.isna()
                & statuses.ne('Active')
            )
        ).any()
    ):
        return 'Payment after last recorded subscription'

    return 'Other unmatched case - review'


# Classify every unmatched payment
unmatched_review['reason_category'] = unmatched_review.apply(classify_unmatched_payment,axis=1)

# Display classification counts
unmatched_category_summary = (
    unmatched_review['reason_category']
    .value_counts()
    .rename_axis('reason_category')
    .reset_index(name='payment_count')
)

display(unmatched_category_summary)

print("Total unmatched payments:", len(unmatched_review))
print("Customers represented:",unmatched_review['customer_id'].nunique())
```
